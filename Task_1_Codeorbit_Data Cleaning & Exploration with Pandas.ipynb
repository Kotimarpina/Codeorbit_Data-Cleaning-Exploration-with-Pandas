```python
import pandas as pd
import numpy as np
print("Libraries imported successfully!")
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
print("Titanic dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
df.info()
```

```python
# Generate descriptive statistics for numerical columns
df.describe()
```

```python
# Calculate the total number of missing values in each column
missing_values = df.isnull().sum()
print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Calculate the percentage of missing values in each column
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("Missing Value Percentage:")
print(missing_percentage.round(2))
```

```python
# Count the total number of duplicate rows
duplicate_count = df.duplicated().sum()
print("Total Duplicate Records:", duplicate_count)
```

```python
# Display all duplicate rows for inspection
df[df.duplicated()]
```

```python
# Remove duplicate rows from the dataset
df = df.drop_duplicates()

# Reset the index after removing duplicates
df = df.reset_index(drop=True)

# Display the updated dataset dimensions
print("Dataset Shape After Removing Duplicates:", df.shape)
```

```python
# Display the data type of every column
print("Column Data Types:")
print(df.dtypes)

```

```python
# Identify numerical columns
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

# Identify categorical columns
categorical_columns = df.select_dtypes(include=['object']).columns

# Display numerical columns
print("Numerical Columns:")
print(list(numerical_columns))

# Display categorical columns
print("\nCategorical Columns:")
print(list(categorical_columns))
```

```python
# Calculate the median age
median_age = df['Age'].median()

# Replace missing Age values with the median
df['Age'] = df['Age'].fillna(median_age)

# Display the remaining missing Age values
print("Missing Age Values:", df['Age'].isnull().sum())
```

```python
# Find the most frequently occurring Embarked value
embarked_mode = df['Embarked'].mode()[0]

# Replace missing Embarked values with the mode
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

# Display the remaining missing Embarked values
print("Missing Embarked Values:", df['Embarked'].isnull().sum())
```

```python
# Calculate the average age of passengers
mean_age = df['Age'].mean()

# Display the mean age
print("Mean Age:", round(mean_age, 2))
```

```python
# Calculate the middle value of passenger ages
median_age = df['Age'].median()

# Display the median age
print("Median Age:", median_age)
```

```python
# Count the number of passengers in each gender category
gender_counts = df['Sex'].value_counts()

# Display the gender distribution
print("Passenger Gender Distribution:")
print(gender_counts)
```

```python
# Count passengers based on their survival status
survival_counts = df['Survived'].value_counts()

# Display survival counts
print("Survival Distribution:")
print(survival_counts)
```

```python
# Check remaining missing values
print("Missing Values:")
print(df.isnull().sum())

# Check remaining duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# Display final dataset dimensions
print("\nFinal Dataset Shape:", df.shape)

# Display updated data types
print("\nFinal Data Types:")
print(df.dtypes)
```

```python
# Replace missing values in the Cabin column with 'Unknown'
df['Cabin'] = df['Cabin'].fillna('Unknown')

# Display the remaining missing values in Cabin
print("Missing Cabin Values:", df['Cabin'].isnull().sum())

# Display the updated Cabin column values
print(df['Cabin'].value_counts().head(10))
```

```python
# Calculate missing values across all columns
missing_values = df.isnull().sum()

# Display the missing-value count
print("Missing Values in Each Column:")
print(missing_values)

# Calculate total missing values
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Check remaining missing values
print("Missing Values:")
print(df.isnull().sum())

# Check remaining duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# Display final dataset dimensions
print("\nFinal Dataset Shape:", df.shape)

# Display updated data types
print("\nFinal Data Types:")
print(df.dtypes)
```

```python
# Save the cleaned dataset as a new CSV file
df.to_csv("../Codeorbit_Data Cleaning & Exploration with Pandas/Titanic_Cleaned.csv", index=False)

# Display confirmation
print("Cleaned dataset exported successfully!")
```
